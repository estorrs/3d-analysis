```python
import os
from pathlib import Path

import numpy as np
import pandas as pd

from io import StringIO
```

## Table S8. 2D automated marker gene discovery: section-level feature correlations

```python
datas = []
for dtype in ['visiumhd', 'xenium', 'cosmx']:
    output_dir = Path(f'../data/spatial/regions/{dtype}')
    for rtype in ['regions', 'tme']:
        corrs = pd.read_csv(output_dir / f'correlations_{rtype}.txt', sep='\t', index_col=0)
        corrs['region_type'] = rtype
        corrs['data_type'] = dtype
        corrs['test_type'] = 'spearman'
        corrs['section_id'] = corrs.index.to_list()
        datas.append(corrs)
corr_df = pd.concat(datas)
corr_df
```

```python
output_dir = Path(f'../data/spatial/regions/visium')
datas = []
for disease in ['breast', 'prostate']:
    vals = pd.read_csv(output_dir / f'ttest_stat_sectionlevel_{disease}.txt', sep='\t', index_col=0)
    vals['section_id'] = vals.index.to_list()
    datas.append(vals)
ttest_df = pd.concat(datas)
ttest_df['region_type'] = 'regions'
ttest_df['data_type'] = 'visium'
ttest_df['test_type'] = 'ttest'
ttest_df
```

```python
merged = pd.concat((corr_df, ttest_df))
merged = merged.melt(id_vars=['section_id', 'region_type', 'data_type', 'test_type'])
merged
```

```python
merged.to_csv('../data/spatial/supplemental/supptable8_feature_metrics.txt.gz', sep='\t', index=False)
```
